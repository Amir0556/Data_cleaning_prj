```python
# inspecting basic structure of data
df.isna().sum() * 100 / df.shape[0]
```

```python
# inspecting unique values in all columns except Transaction Id
df_cols = set(df.columns)
exclude = {'Transaction ID'}
df_cols = df_cols.difference(exclude)
df_cols = list(df_cols)

for col in df_cols:
   print(f'{col}:\n{df[col].unique()}', end = '\n\n')
```

```python
# replacing ERROR and UNKNOWN values with NaN
df = df.replace(['ERROR','UNKNOWN'],np.nan)
df.head(10)
```

```python
# changing data types
df['Quantity'] = df['Quantity'].astype('Int64')
df[['Price Per Unit', 'Total Spent']] = df[['Price Per Unit', 'Total Spent']].astype('float')
df['Transaction Date'] = pd.to_datetime(df['Transaction Date'])
```

```python
#rechecking missing data after treatment
df.isna().sum() * 100 /df.shape[0]
```

```python
# dropping data where transaction date is missing
df = df.dropna(subset = ['Transaction Date'])
```

```python
# replacing misiing values in Payment Method and Location with INA (Informtion Not Available)
# dropping these highvolume records would lead to significant data loss 
#imputing INA avoids bias 
df[['Payment Method','Location']] = df[['Payment Method','Location']].fillna('INA')
```

```python
df.isna().sum() * 100 /df.shape[0]
```

```python
# inspecting mean and median values for Quantity
quantity_mean,quantity_median = round(df['Quantity'].mean(),0),round(df['Quantity'].median(),0)

print(f'Mean Quantity : {quantity_mean}\nMedian Quantity : {quantity_median}')
```

```python
#since mean and median values are same replacing missing values in Quantity with mean value
df['Quantity'] = df['Quantity'].fillna(quantity_mean)
```

```python
#replacing missing values in Item with mode values 
df['Item'] = df['Item'].fillna(df['Item'].mode()[0])
```

```python
# checking how many distinct prices exist for each item category and how many times each price is registered
df.groupby('Item')['Price Per Unit'].value_counts()
```

```python
# filling missing Pricee Per Unit values with most  frequent (mode) price for each Itemcategory 
df['Price Per Unit'] = df.groupby('Item')['Price Per Unit'].transform(lambda x: x.fillna(x.mode()[0]))
```

```python
#filling missing records for Total Spent by calculating the product of quantity times Price Per Unit 
df['Total Spent'] = df['Total Spent'].fillna(df['Quantity'] * df['Price Per Unit'])
```

```python
# storing clean data
df.to_csv('/workspaces/Data_cleaning_prj/data/clean/Cafe Sales(clean).csv ' , index = False)
```
